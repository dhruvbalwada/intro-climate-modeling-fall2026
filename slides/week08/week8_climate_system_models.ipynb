## The climate system and climate models

Adapted from [The Climate Laboratory](https://brian-rose.github.io/ClimateLaboratoryBook) by [Brian E. J. Rose](https://www.atmos.albany.edu/facstaff/brose/), University at Albany.

### Five basic types of climate model

1. Energy Balance Models (EBMs)
2. Radiative-convective (RC) and single-column models (SCMs)
3. Other simplified and integrated modeling approaches
4. Global Circulation Models (GCMs)
5. Earth System Models (ESMs)

### 1. Energy Balance Models (EBMs)

**Energy Balance Model**

A simplified climate model (typically zero- or one-dimensional) predicting the surface temperature as a function of the Earth's energy balance (i.e., the imbalance between absorbed shortwave radiation and outgoing longwave radiation).

Simplified relationships are used to calculate the terms contributing to the energy balance in each latitude zone in the one-dimensional case.

Transport cools the tropics and warms the poles. At equilibrium, transport heating balances net radiation locally.

### 2. Radiative-convective and single-column models

Both resolve the **vertical column**, but the terms describe different aspects of a model.

- **Radiative-convective (RC) model:** focuses on radiation and convection and how they establish a vertical temperature profile and lapse rate.
- **Single-column model (SCM):** runs a GCM’s column physics, including radiation, convection, clouds, turbulence, and surface exchanges. Large-scale transport is prescribed or omitted; the column does not predict a surrounding circulation.

### 3. Other simplified and integrated modeling approaches

Also called **“dimensionally constrained models.”**

- **Statistical-dynamical (SD) models:** calculate a simplified mean atmospheric state, often using latitude and height. Individual weather systems are not resolved; their average effects, such as heat and moisture transport by storms, are represented through statistical relationships. The mean circulation and these statistical effects interact. 
- Earth-System Models of Intermediate Complexity (EMICs) extend SD models to include interactive chemistry, especially the cycling of carbon between ocean, atmosphere and land.
- Integrated Assessment Models (IAMs) couple the climate system to models of economic activity to more fully assess the impact of particular policy choices affective emissions.

### 4. General Circulation Models (GCMs)

GCM can mean **General Circulation Model**, **Global Climate Model**, or **Global Circulation Model**.

- Simulate the **three-dimensional, time-evolving circulation** of the atmosphere, ocean, or both.
- **AGCM:** atmospheric model. **OGCM:** ocean model. **AOGCM / CGCM:** coupled atmosphere–ocean model.
- Originally developed to explain the mean circulation: for example, why surface winds tend to be westerly in midlatitudes and easterly in the tropics.
- Combine circulation with radiation and other physical processes to calculate the evolving climate state. Atmosphere and ocean components can also be run separately for experiments and evaluation.
- Vertical grid spacing is typically much smaller than horizontal spacing, while the number of vertical layers is usually smaller than the number of horizontal columns.

### 5. Earth System Models

Ascending the pyramid even more and moving forward in climate modeling history, more processes that used to be fixed come to be incorporated into GCMs and the coupling becomes more complete: e.g. changes in biomes and vegetation, chemical changes in atmosphere, ocean and soil.  

Such models are now generally known as **Earth System Models**. 

For brevity we use the generic term GCM to mean any complex three-dimensional model of the atmosphere and ocean and possibly other components, used for climate simulation. We will be more precise where necessary.

### Simple vs. Complex models: their different roles in the science

| Aspect | Complex climate models | Simple climate models |
|:---|:---|:---|
| Spatial output | Geographical patterns of temperature, rainfall, winds, and other fields | Mainly global or zonal averages; fewer variables |
| Variability | Can simulate variability such as El Niño and changes in it | Often focus on the mean climate-change signal |
| Process detail | More processes represented; short time steps can resolve the daily cycle | Net effects approximated at coarser spatial and temporal scales |
| Climate sensitivity | Emerges from physical laws and parameterizations | Often specified or calibrated; easy to vary in sensitivity tests |
| Computational cost | Limits the number of experiments and length of spin-up | Enables many scenarios, parameter sweeps, and long integrations |
| Scientific uses | Investigate processes the model can resolve | Explore large-scale interactions and sensitivity |
| Interpretation | Many interacting processes make behavior harder to diagnose | Fewer processes make mechanisms easier to isolate; both model types inform each other |
| Ocean circulation | Coupled models can represent major changes, with uncertainty in timing and character | A 1D model cannot represent circulation changes; added dimensions can help |

*Condensed from The Climate Modelling Primer, Table 2.1, based on Harvey et al. (1997). [Full original table](assets/Primer_Table2.1.png).*

## 4. Introducing the GCM

### Ingredients of a GCM

Predict the **three-dimensional, evolving circulation** of the atmosphere and ocean, together with temperature and transported quantities.

The model solves equations expressing:

- **Momentum:** forces change winds and currents.
- **Mass:** track air, water, and the substances they carry.
- **Energy:** heating, cooling, and transport change temperature.
- **An equation of state:** link pressure, density, and temperature.

Resolved flow is combined with **parameterizations** of processes such as convection, clouds, and turbulent mixing.

### What do the governing equations look like?

A schematic set for a **dry atmosphere**

| Physical law | Equation | What it says |
|:---|:---|:---|
| Momentum | $\displaystyle \frac{D\mathbf{u}}{Dt}+2\boldsymbol{\Omega}\times\mathbf{u}=-\frac{1}{\rho}\nabla p+\mathbf{g}+\mathbf{F}$ | Forces change the velocity of moving air. |
| Mass | $\displaystyle \frac{D\rho}{Dt}=-\rho\nabla\!\cdot\mathbf{u}$ | Expansion lowers density; compression raises it. |
| Energy | $\displaystyle c_p\frac{DT}{Dt}-\frac{1}{\rho}\frac{Dp}{Dt}=Q$ | Heating and pressure changes alter temperature. |
| Equation of state | $\displaystyle p=\rho R_dT$ | Pressure, density, and temperature are linked. |

$D/Dt$ follows moving air. $\mathbf{u}$: velocity; $\rho$: density; $p$: pressure; $T$: temperature.

<small>$\boldsymbol{\Omega}$: Earth’s rotation; $\mathbf{g}$: gravity; $\mathbf{F}$: friction/mixing acceleration; $Q$: diabatic heating per unit mass. Moisture and other tracers require additional equations; model formulations use different approximations.</small>

### A million-dollar question about fluid motion

Our momentum equation belongs to the **Navier–Stokes family**, adapted for a rotating atmosphere. Climate models also include thermodynamics, moisture, and other processes.

- **The Millennium Prize problem:** for specified three-dimensional incompressible flows, prove that smooth solutions persist, or demonstrate finite-time breakdown.
- **September 2026:** closely timed announcements included **Tristan Buckmaster and Levent Alpöge’s** smooth-forced Euler blowup results and **OpenAI’s** announced smooth-forced Navier–Stokes blowup proof. These are related but distinct results. Clay acknowledged an apparent resolution of its problem; evaluation continues.

**The mathematical question is whether smooth fluid motion can break down—even with viscosity and smooth forcing.**

<small>Status checked October 28, 2026. <a href="https://www.claymath.org/news/navier-stokes-announcement/">Clay Mathematics Institute announcement</a> · <a href="https://cims.nyu.edu/~tristanb/statement.pdf">Buckmaster and Alpöge’s work</a> · <a href="https://openai.com/index/navier-stokes-solution/">OpenAI announcement and proof links</a></small>

### What does a singularity mean for numerical modeling?

**A possible singularity does not prevent useful numerical solutions for flows and time intervals where the solution remains smooth and adequately resolved.**

- **Why the mathematics matters:** regularity theory helps establish conditions under which numerical approximations converge and error estimates apply.
- **Near a singularity:** increasingly small scales or steep gradients may require finer grids and smaller time steps. A computer continuing to produce numbers does not guarantee an accurate solution.
- **What a blowup proof establishes:** under specified conditions, smooth initial data and viscosity need not guarantee smooth motion for all time. It does not supply a general method for computing through that breakdown.
- **For climate models:** such a construction does not show that ordinary weather develops singularities or that climate simulations fail. Numerical accuracy and the physical approximations still need evaluation.

<small>Navier–Stokes treats fluid as a continuum; additional physics would matter before a real fluid attained an infinite speed. <a href="https://openai.com/index/navier-stokes-solution/">Announced result and assumptions</a> · <a href="https://www.claymath.org/news/navier-stokes-announcement/">Clay’s evaluation process</a>.</small>

The physical laws are expressed in terms of **differential equations**. 

Just as we saw with the zero-dimensional EBM, we need to approximate the derivatives in order to represent them on a computer.

One way to do this is to divide the Earth into a **series of boxes on a regular grid**. 

The model then represents all the **state variables**, e.g. for the atmosphere

- Temperature
- Wind speed and direction
- Humidity
- Pressure

for each box - essentially a spatial average of these quantities over the box.

The discretized equations are then solved numerically by **timestepping**, (conceptually at least) just like we have already done with the EBM.

### Discretizing space: three approaches

The same physical laws can have different numerical representations.

| Approach | Main idea | What to notice |
|:---|:---|:---|
| **Finite differences** | Approximate derivatives using nearby grid values: $\partial T/\partial x\approx(T_{i+1}-T_{i-1})/(2\Delta x)$. | The choice of neighboring points affects accuracy and behavior. |
| **Finite volumes** | Evolve cell averages from fluxes crossing cell faces. | The outflow from one cell is the inflow to its neighbor: budgets can be conserved. |
| **Spectral methods** | Represent a field as a sum of basis functions, such as waves or spherical harmonics. | Derivatives come from the basis functions; smooth fields can be represented accurately. |

**Lab connection:** our latitude-band transport budget uses shared boundary fluxes, the finite-volume idea.

<small>Examples: <a href="https://www.gfdl.noaa.gov/fv3/fv3-key-components/">GFDL finite-volume core</a> · <a href="https://www.gfdl.noaa.gov/atmospheric-model/">GFDL spectral models</a>. Models may combine methods in different directions or processes.</small>

### Same advection equation, different numerical behavior

A warm patch carried by a constant wind: $\partial T/\partial t+u\,\partial T/\partial x=0$. **No physical diffusion.**

![Exact transport compared with numerical spreading from upwind and oscillations from Lax–Wendroff](assets/advection_schemes.png)

- **Upwind:** numerical diffusion spreads the patch’s edges.
- **Lax–Wendroff:** a second-order method keeps sharper edges here, but creates artificial oscillations.

**Both conserve the total anomaly here. Conservation alone does not guarantee an accurate shape.**

### Representing forcing, feedbacks, and heat uptake

A comprehensive model calculates how **external changes and internal responses alter the planetary energy budget**.

**Change CO₂ → calculate radiative transfer → diagnose forcing → simulate the evolving climate.**

- **Forcing $F$:** the radiative-budget change caused by an imposed driver before global surface warming, with specified adjustments allowed. A concentration change in ppm is not itself a forcing in W m⁻².
- **Feedbacks $\lambda$:** the radiative response as temperature, water vapor, clouds, lapse rate, and albedo change with warming.
- **Heat uptake:** the evolving energy imbalance changes heat storage, mainly in the ocean, and affects the pace of surface warming.

<small>Forcing definitions distinguish instantaneous effects from rapid adjustments. In a concentration-driven experiment, CO₂ is prescribed and radiative fluxes are calculated. <a href="https://brian-rose.github.io/ClimateLaboratoryBook/courseware/sensitivity-feedback/">Rose: forcing and feedback calculations</a>.</small>

### Comprehensive models need parameterizations

Comprehensive models cannot explicitly simulate every process at every scale. Parameterizations represent the effects of unresolved processes using physical relationships and evidence from observations or detailed models.

Land surface processes are a good example.

Exchanges of energy, water and carbon between the land and atmosphere are biologically mediated. We must (to a certain extent) rely on empirical relationships. A bit like economic modeling.

We also must deal with interaction across spatial scales.

E.g. cumulus convection and vertical transport of heat and water vapor
