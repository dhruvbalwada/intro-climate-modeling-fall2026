From the board: a bucket with water flowing in and draining out. Budget of water volume ($V = A\,h$):

$$A\,\frac{dh}{dt} \;=\; \underbrace{F}_{\text{inflow (forcing — external)}} \;-\; \underbrace{k\,h}_{\text{drain (property of the system)}}$$

- $h$ = water level [m], $A$ = bucket area [m²]
- $F$ = inflow [m³/s] — someone pours; the bucket doesn't decide
- $k h$ = outflow [m³/s] — faster when fuller: set by the bucket itself ($k$ [m²/s])

## 1. Equilibrium — no time needed

At equilibrium the level isn't changing: $dh/dt = 0$, so inflow = outflow:

$$h^* = F / k$$

**Check the units. Then compute it:**

```python
import numpy as np
import matplotlib.pyplot as plt

A = 1.0    # m^2
F = 2.0    # m^3/s
k = 1.0    # m^2/s

h_star = F / k
print(f"equilibrium level h* = {h_star} m")
```

## 2. Step forward in time

The equation tells us the *rate of change* right now. So predict a small step ahead, over and over:

$$h_{\text{next}} = h_{\text{now}} + \Delta t \cdot \frac{F - k\,h_{\text{now}}}{A}$$

*(This is called Euler's method — it's how almost every model marches time forward.)*

The level rises and *settles at the equilibrium* — the drain term pulls it there (fuller → drains harder → a **stabilizing feedback**).
